```python
#讓使用者輸入密碼，若密碼為 1234 則輸出歡迎光臨，否則提示密碼錯誤：

password = input("請輸入密碼:")
if password == "1234":
    print("密碼正確!歡迎光臨!")
else:
    print("密碼錯誤\n請重新輸入")
```

```python
#巢狀選擇 (Nested if)
furry = False
small = True

if furry:
    if small:
        print("它是貓!")
    else:
        print("它是熊!")
else:
    if small:
        print("它是小蜥蜴")
    else:
        print("它是人類或是沒毛的熊")
        
```
